```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)

df = iris.frame
df["species"] = df["target"].replace({0: "setosa",1: "versicolor",2: "virginica"})

mean_petal = df["petal length (cm)"].mean()

result = df[df["petal length (cm)"] > mean_petal][["species", "petal length (cm)"]]

print(result)
print("Rows:", len(result))
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)

df = iris.frame
df["petal_area"] = (df["petal length (cm)"]*df["petal width (cm)"])
df["is_big"] = df["petal_area"] > 10

print(df["is_big"].value_counts())
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)

df = iris.frame
df["species"] = df["target"].replace({0: "setosa",1: "versicolor",2: "virginica"})

result = df.groupby("species").mean()

print(result)
print(result["sepal length (cm)"].idxmax())
```
