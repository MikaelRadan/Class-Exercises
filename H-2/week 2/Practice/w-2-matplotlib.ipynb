```python
import matplotlib.pyplot as plt

epoch_losses = [0.9,0.6,0.45,0.38,0.34,0.31,0.30]

fig,ax = plt.subplots()

ax.plot(epoch_losses)
ax.set_title("Training Loss")
ax.set_xlabel("epoch")
ax.set_ylabel("loss")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ["cat","dog","cat","cat","bird","dog"]

unique_labels, counts = np.unique(labels,return_counts=True)

fig,ax = plt.subplots()
ax.bar(unique_labels,counts)
ax.set_title("Category Counts")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()
data = rng.normal(size=(2000,4))
fig,axes = plt.subplots(2,2)

for i,ax in enumerate(axes.ravel()):
    ax.hist(data[:,i])
    ax.set_title(f"Feature {i+1}")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.array([1,2,3,4,5,6])
x2 = np.array([2,1,4,3,6,5])

y_true = np.array([1,0,1,1,0,1])
y_pred = np.array([1,1,1,0,0,1])

correct = (y_true == y_pred)

fig,ax = plt.subplots()

ax.scatter(x1[correct],x2[correct],label="correct")
ax.scatter(x1[~correct],x2[~correct],label="incorrect")
ax.legend()
ax.set_title("Prediction Results")

plt.show()
```
