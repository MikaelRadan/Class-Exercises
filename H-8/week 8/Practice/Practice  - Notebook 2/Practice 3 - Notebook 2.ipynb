## Wine: PCA with 2 Components Instead of 8

The task from the course notebook is: **Which wines get flagged with 2 components instead of 8? Are they the same ones?**

We follow the course notebook's method: standardize the Wine features, calculate PCA reconstruction error, and flag the top 2% of scores.

## Cell 1 — Import the Required Libraries

Import NumPy, Matplotlib, `load_wine`, `StandardScaler`, and `PCA`.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Cell 2 — Load and Standardize the Wine Dataset

The dataset contains 178 wines. Standardize the features before applying PCA, as in the course notebook.

```python
wine = load_wine(as_frame=True)
X_wine = wine.data
X_wine_scaled = StandardScaler().fit_transform(X_wine)

print("Dataset shape:", X_wine.shape)
print("Number of features:", X_wine.shape[1])
```

## Cell 3 — Calculate Reconstruction Error with 8 Components

First, reproduce the course notebook's approach with 8 PCA components. The reconstruction error is the mean squared difference between the standardized data and its reconstruction.

```python
pca_8 = PCA(n_components=8).fit(X_wine_scaled)
X_wine_back_8 = pca_8.inverse_transform(pca_8.transform(X_wine_scaled))
wine_scores_8 = ((X_wine_scaled - X_wine_back_8) ** 2).mean(axis=1)

threshold_8 = np.percentile(wine_scores_8, 98)
flagged_8 = np.where(wine_scores_8 > threshold_8)[0]

print("8 components")
print("threshold:", round(threshold_8, 3))
print("flagged wines:", flagged_8)
```

## Cell 4 — Repeat with 2 Components

Now change only the number of PCA components from 8 to 2. Keep the same reconstruction-error calculation and the same 98th-percentile threshold.

```python
pca_2 = PCA(n_components=2).fit(X_wine_scaled)
X_wine_back_2 = pca_2.inverse_transform(pca_2.transform(X_wine_scaled))
wine_scores_2 = ((X_wine_scaled - X_wine_back_2) ** 2).mean(axis=1)

threshold_2 = np.percentile(wine_scores_2, 98)
flagged_2 = np.where(wine_scores_2 > threshold_2)[0]

print("2 components")
print("threshold:", round(threshold_2, 3))
print("flagged wines:", flagged_2)
```

## Cell 5 — Compare the Results

Compare the two sets of flagged row indices and identify any wines flagged in both cases.

```python
common_wines = np.intersect1d(flagged_2, flagged_8)

print("Flagged with 2 components:", flagged_2.tolist())
print("Flagged with 8 components:", flagged_8.tolist())
print("Flagged in both cases:", common_wines.tolist())
print("Are the sets identical?", np.array_equal(np.sort(flagged_2), np.sort(flagged_8)))
```

## Final Answer

- **With 8 components:** wines at row indices **73, 84, 105, and 121** are flagged.
- **With 2 components:** wines at row indices **59, 69, 95, and 121** are flagged.
- **Are they the same?** No. Only row index **121** is flagged in both cases.

Using only 2 components preserves less information, so the reconstructed data differs more from the original data. This changes the reconstruction-error scores and which wines are flagged. These are zero-based row indices in the dataset.
