## Wine: Find the Three Features with the Largest Weights in PC2

In this exercise, we use the standardized Wine dataset and examine the **second principal component (PC2)**.

Each feature has a weight (loading) in PC2. We will find the three features with the largest absolute weights, considering both positive and negative values, and then describe PC2 in words.

**Question:** Which three features have the largest weights in PC2, and how would you describe PC2?

## Cell 1 — Import the Required Libraries

We will use pandas to display the feature weights, `load_wine` to load the dataset, `StandardScaler` to standardize the features, and `PCA` to calculate the principal components.

```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Cell 2 — Load the Wine Dataset

The Wine dataset contains 178 samples, 13 chemical features, and three wine classes.

```python
wine = load_wine(as_frame=True)

X = wine.data
y = wine.target

print("Dataset shape:", X.shape)
print("Number of features:", X.shape[1])
print("Number of classes:", y.nunique())
```

## Cell 3 — Standardize the Features

We standardize the features before PCA so that features measured on different scales can contribute appropriately.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Standardized data shape:", X_scaled.shape)
```

## Cell 4 — Fit PCA

Fit PCA to the standardized data. The feature weights for PC2 are stored in the second row of `pca.components_`.

The sign of a weight indicates its direction. To find the strongest weights whether positive or negative, we sort by **absolute value**.

```python
pca = PCA()
pca.fit(X_scaled)

pc2_weights = pd.Series(
    pca.components_[1],
    index=X.columns,
    name="PC2 weight"
)

print("PC2 explained variance:", round(pca.explained_variance_ratio_[1] * 100, 2), "%")
pc2_weights.sort_values(key=abs, ascending=False).to_frame()
```

## Cell 5 — Find the Three Largest Weights

Sort the PC2 weights by absolute value and select the first three. The original signed weights are kept so we can see whether each feature contributes positively or negatively to PC2.

```python
top_3_pc2 = pc2_weights.loc[
    pc2_weights.abs().sort_values(ascending=False).head(3).index
]

print("Three features with the largest absolute weights in PC2:")
display(top_3_pc2.to_frame())
```

## Final Answer

The three features with the largest absolute weights in PC2 are:

1. **Color intensity:** `+0.530`
2. **Alcohol:** `+0.484`
3. **Proline:** `+0.365`

These are the three strongest feature weights by absolute value. In this PCA result, all three weights are positive, so higher values for these features contribute in the positive direction along PC2.

**Interpretation:** PC2 is influenced most strongly by color intensity, alcohol, and proline. It can be described as a component that contrasts wine samples according to a combination of these properties. PC2 combines all 13 standardized features, so this interpretation focuses on the three strongest weights rather than the entire component.

**Note:** The overall sign of a principal component can be flipped by PCA without changing the underlying result. Therefore, the relative weights and their magnitudes are more important than whether all signs appear positive or negative.
