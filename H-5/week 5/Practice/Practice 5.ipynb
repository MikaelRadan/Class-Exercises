Compare GaussianNB's test accuracy, precision, and recall (notebook 4, average="macro") against Random Forest's (notebook 1). How large is the gap on this image dataset? Would you expect it to be larger or smaller on a text-classification dataset, and why?

Cell 1 — Import the required libraries

In this cell, we only enter the libraries and methods that are used in the Master Model Comparison structure.

```python
# Import the dataset, models, scaler, and evaluation metrics.

import pandas as pd
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

Cell 2 — Load the Digits dataset

We use the same Digits dataset

```python
# Load the Digits dataset and separate features from the target.

data = load_digits(as_frame=True)
df = data.frame

X = df.drop(columns="target")
y = df["target"]
```

Cell 3 - Split the data into training and testing sets

The split is performed: 80% training and 20% testing, with stratify=y.

```python
# Split the data while preserving the class distribution.

X_train, X_test, y_train, y_test = train_test_split(
    X ,y , test_size=0.2 , random_state=13 , stratify=y)
```

Cell 4 - Scale the features

The data was scaled using StandardScaler prior to training.

```python
# Fit the scaler on training data and transform both datasets.

scaler = StandardScaler()

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Cell 5 — Train Gaussian Naive Bayes

Now, we train the Gaussian Naive Bayes model.

```python
# Train Gaussian Naive Bayes and generate test predictions.

nb = GaussianNB()

nb.fit(X_train_scaled, y_train)

y_pred_nb = nb.predict(X_test_scaled)
```

Cell 6 — Train Random Forest

```python
# Train Random Forest with the same settings used by the instructor.

forest = RandomForestClassifier(
    n_estimators=200 , random_state=13)

forest.fit(X_train_scaled, y_train)

y_pred_forest = forest.predict(X_test_scaled)
```

Cell 7 — Compare the two models

```python
# Compare Gaussian Naive Bayes and Random Forest on the test set.

model_predictions = {
    "GaussianNB": y_pred_nb , "Random Forest": y_pred_forest}

comparison = pd.DataFrame({
    name: {
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, average="macro"),
        "Recall": recall_score(y_test, pred, average="macro")}
    for name, pred in model_predictions.items()})

comparison.round(4).T
```

Practice 5 — NB vs. Random Forest

Random Forest clearly outperforms GaussianNB on this image dataset: a 22-point accuracy gap (0.9778 vs. 0.7556) and similarly large gaps in precision and recall.

 This gap is likely to shrink on a text-classification dataset, because Naive Bayes' core assumption — that features are conditionally independent given the class — holds much better for word counts than for pixel intensities, where neighboring pixels are strongly correlated.
 
  Naive Bayes (typically MultinomialNB, not GaussianNB, for word-count data) is a classic strong baseline for text classification, often competitive with far more complex models on high-dimensional sparse text features.
