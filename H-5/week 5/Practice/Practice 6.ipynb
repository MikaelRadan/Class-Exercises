Cell 1 — Import the required libraries

In this cell, we import only the tools necessary to run those eight models and calculate the accuracy.

```python
# Import the dataset, models, preprocessing tools, and accuracy metric.

import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier)
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score
```

Cell 2 — Load the Wine dataset

As per the exercise instructions, we use load_wine(as_frame=True) instead of the Digits image dataset.

This dataset contains 13 chemical features and 3 classes.

```python
# Load the Wine dataset and separate features from the target.

data = load_wine(as_frame=True)
df = data.frame

X = df.drop(columns="target")
y = df["target"]
```

Cell 3 — Split the data

```python
# Split the data while preserving the class distribution.

X_train, X_test, y_train, y_test = train_test_split(
    X , y , test_size=0.2 , random_state=13 , stratify=y)
```

Cell 4 — Scale the features

First, we fit the StandardScaler to the training data, and then we transform both the Train and Test sets.

```python
# Fit the scaler on training data and transform both datasets.

scaler = StandardScaler()

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Cell 5 — Logistic Regression and KNN

We train the first two models using the same settings specified by the instructor.

```python
# Train Logistic Regression and KNN on the scaled Wine data.

logreg = LogisticRegression(max_iter=5000)
logreg.fit(X_train_scaled, y_train)
y_pred_logreg = logreg.predict(X_test_scaled)

knn = KNeighborsClassifier(n_neighbors=8)
knn.fit(X_train_scaled, y_train)
y_pred_knn = knn.predict(X_test_scaled)
```

Cell 6 — Decision Tree and Random Forest

The Decision Tree retains max_depth=8, and the Random Forest retains n_estimators=200.

```python
# Train Decision Tree and Random Forest with the instructor settings.

tree = DecisionTreeClassifier(max_depth=8, random_state=13)
tree.fit(X_train_scaled, y_train)
y_pred_tree = tree.predict(X_test_scaled)

forest = RandomForestClassifier(n_estimators=200, random_state=13)
forest.fit(X_train_scaled, y_train)
y_pred_forest = forest.predict(X_test_scaled)
```

Cell 7 — AdaBoost and Gradient Boosting

The boosting settings also remain unchanged:

AdaBoost → n_estimators=100
Gradient Boosting → n_estimators=100, learning_rate=0.1

```python
# Train AdaBoost and Gradient Boosting with the instructor settings.

ada = AdaBoostClassifier(n_estimators=100, random_state=13)
ada.fit(X_train_scaled, y_train)
y_pred_ada = ada.predict(X_test_scaled)

gb = GradientBoostingClassifier(
    n_estimators=100 , learning_rate=0.1 , random_state=13)
gb.fit(X_train_scaled, y_train)
y_pred_gb = gb.predict(X_test_scaled)
```

Cell 8 — SVM and Gaussian Naive Bayes

The SVM is exactly the RBF model with C=1 and gamma="scale", and GaussianNB is used without modification.

```python
# Train the RBF SVM and Gaussian Naive Bayes models.

svm_rbf = SVC(kernel="rbf", C=1, gamma="scale")
svm_rbf.fit(X_train_scaled, y_train)
y_pred_svm_rbf = svm_rbf.predict(X_test_scaled)

nb = GaussianNB()
nb.fit(X_train_scaled, y_train)
y_pred_nb = nb.predict(X_test_scaled)
```

Cell 9 — Compare test accuracy

The exercise requires reporting the test accuracy for all eight models; therefore, we calculate only the requested metric.

```python
# Calculate test accuracy for all eight models.

model_predictions = {
    "Logistic Regression": y_pred_logreg,
    "KNN (k=8)": y_pred_knn,
    "Decision Tree": y_pred_tree,
    "Random Forest": y_pred_forest,
    "AdaBoost": y_pred_ada,
    "Gradient Boosting": y_pred_gb,
    "SVM (RBF)": y_pred_svm_rbf,
    "Naive Bayes": y_pred_nb}

accuracy_results = {
    name: accuracy_score(y_test, pred)
    for name, pred in model_predictions.items()}

accuracy_table = pd.DataFrame(
    accuracy_results.items(),
    columns=["Model", "Test Accuracy"])

accuracy_table.round(4).sort_values(
    "Test Accuracy" , ascending=False)
```

Cell 10 — Final interpretation

In this cell, we simply take the comparison result from the table and identify the best model.

```python
# Identify the model with the highest test accuracy.

best_model = accuracy_table.iloc[0]

print("Best model:", best_model["Model"])
print("Test accuracy:", round(best_model["Test Accuracy"], 4))
```
