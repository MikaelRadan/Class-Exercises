```python

import numpy as np

arr = np.random.randint(0,101,(5,5))

print("Before:")
print(arr)

arr[arr > 50] = 0

print("After:")
print(arr)
```

```python
import numpy as np

arr = np.random.randint(0,100,(10,3))
print("Orlginal")
print(arr)

min1 = arr.min(axis=0)
max2 = arr.max(axis=0)

normal = (arr - min1) / (max2 - min1)

print("Normalized:")
print(normal)
```

```python
import numpy as np

scores = np.array([[0.1,0.7,0.2],[0.8,0.1,0.1],[0.2,0.3,0.5],[0.9,0.05,0.05],[0.1,0.2,0.7]])

true_labels = np.array([1,0,2,0,2])

predicted = np.argmax(scores,axis = 1)

accuracy = np.mean(predicted == true_labels)

print("Prediction:")
print(predicted)

print("Accuracy:")
print(accuracy)
```
